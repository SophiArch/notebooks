**Task:** An AI assistant was asked to check whether three datasets are normally distributed and apply appropriate transforms. Review its analysis: identify where it read distribution shapes correctly, where it misdiagnosed, and where it applied unnecessary transforms.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)

# Three datasets to analyse
# A: employee salaries (right-skewed)
salaries = np.concatenate([
    np.random.lognormal(mean=10.8, sigma=0.6, size=800),
    np.random.lognormal(mean=12.5, sigma=0.4, size=200)
])

# B: product ratings (discrete, near-normal for this dataset)
ratings = np.clip(np.round(np.random.normal(3.8, 0.7, 600)).astype(int), 1, 5)

# C: session durations in seconds (bimodal: short visits + long sessions)
durations = np.concatenate([
    np.random.normal(45, 12, 500),
    np.random.normal(280, 40, 500)
])
durations = np.clip(durations, 1, None)

print(f'Salaries: n={len(salaries)}, mean={salaries.mean():.0f}, median={np.median(salaries):.0f}')
print(f'Ratings: n={len(ratings)}, mean={ratings.mean():.2f}, median={np.median(ratings):.1f}')
print(f'Durations: n={len(durations)}, mean={durations.mean():.0f}s, median={np.median(durations):.0f}s')
```

## AI-generated analysis

The AI was given the three datasets and asked: *"Check whether each dataset is normally distributed and apply a log transform to any that aren't."*

Here is its output:

```python
# --- AI-GENERATED CODE ---

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
datasets = {'Salaries': salaries, 'Ratings': ratings, 'Durations': durations}

for i, (name, data) in enumerate(datasets.items()):
    # Original distribution
    axes[0, i].hist(data, bins=40, color='steelblue', edgecolor='white', alpha=0.8)
    axes[0, i].set_title(f'{name} (original)')

    # Log transform (applied to ALL three datasets)
    log_data = np.log(data)
    axes[1, i].hist(log_data, bins=40, color='coral', edgecolor='white', alpha=0.8)
    axes[1, i].set_title(f'{name} (log-transformed)')

plt.tight_layout()
plt.show()

# AI conclusion text (generated as a print statement to simulate its output)
conclusions = {
    'Salaries': 'Right-skewed. Log transform applied. Distribution now approximately normal.',
    'Ratings': 'Slightly asymmetric. Log transform applied as a precaution.',
    'Durations': 'Skewed. Log transform applied. Distribution now approximately normal.'
}
for name, conclusion in conclusions.items():
    print(f'{name}: {conclusion}')
```

## Step 2: Run Shapiro-Wilk normality checks

The AI did not run any formal normality test. Run Shapiro-Wilk on all three *original* datasets.

```python
# Note: Shapiro-Wilk is unreliable for large n (>5000). Sample if needed.
for name, data in [('Salaries', salaries), ('Ratings', ratings.astype(float)), ('Durations', durations)]:
    sample = data if len(data) <= 5000 else np.random.choice(data, 5000, replace=False)
    stat, p = stats.shapiro(sample)
    result = 'NORMAL' if p > 0.05 else 'NON-NORMAL'
    print(f'{name}: W={stat:.4f}, p={p:.4f} → {result}')
```

> **Question 3:** Do the Shapiro-Wilk results confirm or contradict the AI's assessment for each dataset? Note that for very large samples, even minor deviations from normality will be detected — use your judgment about whether the p-value reflects a *meaningful* departure from normality.

*(Write your answer here.)*

**Salaries:** Shapiro-Wilk rejects normality (p < 0.05) — confirms the AI was right to flag it as non-normal, though log-transforming should be verified rather than assumed. **Ratings:** Shapiro-Wilk likely flags non-normality because the data is discrete ordinal (integers 1–5), not a limitation of the distribution's shape — the AI should not have applied a log transform at all, and the test result doesn't change that. **Durations:** Shapiro-Wilk strongly rejects normality because the bimodal structure departs dramatically from a bell curve; this contradicts the AI's claim that the log transform achieved approximate normality. For very large samples, Shapiro-Wilk detects even trivial deviations — use the histogram alongside the p-value before concluding non-normality is practically meaningful.
