## Expected AI failure modes for this prompt

When given a prompt such as "create a bar chart showing Q4 sales by region"
without explicit colour constraints, typical AI output will exhibit several
of the five failure patterns from Lesson 9:

**Near-certain failures (appear in ~90% of naive outputs):**
- **Multi-colour palette on a single series** — AI defaults apply a
  qualitative palette (`tab10`, `Set2`, or explicit colour lists) to the five
  bars, implying a categorical distinction that doesn't exist. Criterion 1.
- **Variable-name title** — output titles are typically "Q4 Sales by Region",
  "Regional Sales Performance", or "Sales by Region (Q4)" — none state a
  finding. Criterion 2.

**Frequent failures (appear in ~60–70% of outputs):**
- **Missing or unit-free axis label** — y-axis is often labelled "Sales"
  with no currency unit, or omitted entirely. Criterion 3.
- **Spines not removed** — default matplotlib style retains all four spines;
  AI rarely removes top and right unprompted. Criterion 4.

**Occasional failures (appear in ~30–40% of outputs):**
- **No savefig call or missing export settings** — when a savefig call is
  present, `facecolor='white'` is the most commonly omitted argument.
  Criterion 5.
- **Unnecessary legend** — some outputs add a legend for a single-series
  chart; there is nothing to distinguish. Not a numbered criterion but worth
  flagging in discussion.
