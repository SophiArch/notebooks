## Completed Pipeline

The cell below shows the full working solution for reference. Review it after completing the tasks above.

```python
from tensorflow.keras.callbacks import EarlyStopping

# Full solution: model with Dropout, compiled, trained with EarlyStopping
model_solution = keras.Sequential([
    layers.Input(shape=(4,)),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(32, activation='relu'),
    layers.Dense(3, activation='softmax')
])

model_solution.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

early_stop_solution = EarlyStopping(patience=10, restore_best_weights=True)

history_solution = model_solution.fit(
    X_train, y_train,
    epochs=100,
    validation_split=0.2,
    callbacks=[early_stop_solution],
    verbose=0
)

test_loss, test_acc = model_solution.evaluate(X_test, y_test, verbose=0)
print(f"Stopped at epoch: {len(history_solution.history['loss'])}")
print(f"Test accuracy: {test_acc:.3f}")
```
