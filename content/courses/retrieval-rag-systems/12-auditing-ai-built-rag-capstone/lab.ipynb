## The audit protocol

Audit each round against these 7 checks, in pipeline order. Each check is testable by inspection — cite the line(s) of pasted code that pass or fail it.

1. **Chunking**: boundaries respect document structure (headings/tables); every chunk carries doc_id + heading path + tenant metadata.
2. **Update path**: document changes are delete-then-insert keyed by doc_id (no chunk-ID upserts that orphan old tails); embedding model version pinned and asserted.
3. **Retrieval**: hybrid where traffic warrants (error codes ⇒ yes here); fusion by RRF over ranks, never weighted score sums; one tokeniser at index and query time.
4. **Funnel**: stage-one k meaningfully wider than final top-k; re-ranking present; a relevance floor for the honest no-answer.
5. **Evaluation**: recall@k / MRR computed correctly, at the k production uses, zero-scoring queries kept; eval queries are realistic phrasings, not document titles.
6. **Permissions**: tenant filter inside the search (or per-tenant namespaces), never after it; no cross-tenant cache keys.
7. **Observability & cost**: per-stage trace with chunk IDs (not text); someone did the input-token arithmetic for top-k × chunk size.
