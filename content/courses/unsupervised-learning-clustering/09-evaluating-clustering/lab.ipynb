## Step 2: Generate Your Data

```python
np.random.seed(42)

# Create three clusters
cluster_1 = np.random.normal([0, 0], 1.5, (50, 2))
cluster_2 = np.random.normal([8, 0], 1.5, (50, 2))
cluster_3 = np.random.normal([4, 7], 1.5, (50, 2))

X = np.vstack([cluster_1, cluster_2, cluster_3])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Run k-means clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X_scaled)

print(f"Data shape: {X_scaled.shape}")
print(f"Cluster sizes: {np.bincount(labels)}")
```

## Step 6: Run the Corrected Code

```python
# After auditing, either run the AI's code as-is (if correct) or fix it below.

# Corrected code (or AI's code if it passed audit):
sil_score = silhouette_score(X_scaled, labels)
db_score = davies_bouldin_score(X_scaled, labels)
ch_score = calinski_harabasz_score(X_scaled, labels)

print(f"Silhouette Score: {sil_score:.3f}")
print(f"Davies-Bouldin Index: {db_score:.3f}")
print(f"Calinski-Harabasz Score: {ch_score:.1f}")

print("\nInterpretation:")
print(f"Silhouette {sil_score:.3f}: Score > 0.5 suggests good separation.")
print(f"Davies-Bouldin {db_score:.3f}: Lower is better; < 1.5 is good.")
print(f"Calinski-Harabasz {ch_score:.1f}: Higher is better; > 100 suggests good separation.")

print("\nDecision:")
if sil_score > 0.5 and db_score < 1.5 and ch_score > 100:
    print("✓ These metrics suggest good clustering quality.")
else:
    print("✗ At least one metric suggests the clustering could be improved.")
    print("  Consider trying different k values or different clustering algorithms.")
```

## Summary

Evaluating unsupervised clustering requires multiple metrics:
- **Silhouette Score:** -1 to 1, higher is better
- **Davies-Bouldin Index:** lower is better
- **Calinski-Harabasz Score:** higher is better

When you have no ground truth, these metrics guide your decision about cluster quality. But remember: no metric is perfect. Always examine the clusters visually and understand them in your domain context.
