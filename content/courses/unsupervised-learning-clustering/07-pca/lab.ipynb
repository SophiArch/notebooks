```python
import numpy as npimport pandas as pdfrom sklearn.preprocessing import StandardScalerfrom sklearn.decomposition import PCAfrom sklearn.cluster import KMeansimport matplotlib.pyplot as plt
```

## Step 1: Generate High-Dimensional Data

```python
# Create a high-dimensional dataset with redundancynp.random.seed(42)n_samples = 200n_informative = 5  # Only a few dimensions carry real signaln_noise = 20       # Rest is noise# Signal dimensionssignal = np.random.normal(0, 2, (n_samples, n_informative))# Noise dimensionsnoise = np.random.normal(0, 1, (n_samples, n_noise))X = np.hstack([signal, noise])scaler = StandardScaler()X_scaled = scaler.fit_transform(X)print(f"Data shape: {X_scaled.shape}")print(f"True dimensions: {n_informative} signal + {n_noise} noise")
```

## Step 2: Baseline — Fit PCA and Visualize Variance

```python
# Baseline code: fit PCA and show explained variancepca = PCA()pca.fit(X_scaled)# Variance explained by each componentcumulative_variance = np.cumsum(pca.explained_variance_ratio_)plt.figure(figsize=(10, 5))plt.subplot(1, 2, 1)plt.bar(range(1, len(pca.explained_variance_ratio_) + 1),        pca.explained_variance_ratio_)plt.xlabel('Principal Component')plt.ylabel('Variance Explained')plt.title('Variance Explained per Component')plt.subplot(1, 2, 2)plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance)plt.xlabel('Number of Components')plt.ylabel('Cumulative Variance Explained')plt.title('Cumulative Variance Explained')plt.axhline(y=0.95, color='r', linestyle='--', label='95% threshold')plt.legend()plt.tight_layout()plt.show()n_components_95 = np.argmax(cumulative_variance >= 0.95) + 1print(f"\nComponents needed for 95% variance: {n_components_95}")print(f"Original dimensions: {X_scaled.shape[1]}")print(f"Reduction: {X_scaled.shape[1]} -> {n_components_95}")
```

Does it preserve the structure you care about?- PCA before a supervised model: Caution.
